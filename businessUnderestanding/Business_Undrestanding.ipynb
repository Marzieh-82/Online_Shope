### سناریو تسک فروشگاه آنلاین



چالش پیشرفته داده‌کاوی: تحلیل رفتار مشتریان و شناسایی فرصت‌های بازاریابی

پروژه تحلیل رفتار خرید، بخش‌بندی مشتریان و طراحی راهبرد حفظ و توسعه مشتری

۱. سناریوی کسب‌وکار (Business Understanding)

شما به‌عنوان تحلیلگر ارشد داده‌کاوی در یک فروشگاه آنلاین بین‌المللی فعالیت می‌کنید.

مدیریت فروشگاه متوجه شده است که مشتریان رفتارهای خرید متفاوتی دارند. برخی مشتریان به‌صورت مستمر خرید می‌کنند و ارزش اقتصادی بالایی برای کسب‌وکار ایجاد می‌کنند؛ برخی خریدهای پراکنده دارند؛ تعدادی از مشتریان در گذشته فعال بوده‌اند اما اخیراً فعالیت خرید آن‌ها کاهش یافته است و گروهی نیز با وجود سابقه خرید، فرصت‌های اقتصادی قابل‌توجهی برای فروشگاه ایجاد نمی‌کنند.

شرکت قصد دارد با تحلیل سوابق تراکنش مشتریان:

1. الگوهای مختلف رفتار خرید را شناسایی کند.
2. مشتریان را بر اساس شباهت رفتاری در گروه‌های مشخص قرار دهد.
3. ارزش اقتصادی هر مشتری را ارزیابی کند.
4. مشتریانی را که فرصت بیشتری برای بازاریابی و بازگرداندن آن‌ها وجود دارد، شناسایی کند.
5. با توجه به محدودیت بودجه، ۵۰۰ مشتری را برای اجرای کمپین بازاریابی انتخاب و اولویت‌بندی کند.
6. با بررسی الگوهای خرید محصولات، پیشنهادهای مرتبط و هدفمند برای مشتریان طراحی کند.

چالش اصلی پروژه:

دیتاست خام فروشگاه فاقد برچسب آماده‌ای برای ریزش مشتری است. بنابراین امکان استفاده مستقیم از یک مدل طبقه‌بندی برای تشخیص مشتریان ریزشی و غیرریزشـی وجود ندارد.

در این پروژه، به‌جای ساخت یک مدل پیش‌بینی ریزش، از تحلیل توصیفی رفتار مشتریان، الگوریتم‌های خوشه‌بندی و معیارهای ارزش اقتصادی و فرصت بازاریابی استفاده می‌شود تا مشتریان مناسب برای اقدامات بازاریابی شناسایی شوند.

هدف نهایی، ایجاد یک سامانه تحلیلی برای پشتیبانی از تصمیم‌های بازاریابی و تخصیص بهینه منابع شرکت است.

۲. دیتاست

منبع: UCI Online Retail

لینک:
https://archive.ics.uci.edu/dataset/352/online+retail

دیتاست شامل تراکنش‌های یک فروشگاه آنلاین بین‌المللی در بازه زمانی دسامبر ۲۰۱۰ تا دسامبر ۲۰۱۱ است.

ستون‌های اصلی:

- InvoiceNo: شماره فاکتور
- StockCode: کد محصول
- Description: توضیحات محصول
- Quantity: تعداد کالا
- InvoiceDate: تاریخ تراکنش
- UnitPrice: قیمت واحد
- CustomerID: شناسه مشتری
- Country: کشور مشتری

این داده‌ها شامل خریدهای مشتریان، فاکتورهای لغوشده، مقادیر منفی، مقادیر گمشده و الگوهای رفتاری متفاوت هستند.

۳. مرحله اول: ممیزی و شناخت کیفیت داده‌ها (Data Understanding)

پیش از هرگونه تحلیل، کیفیت و ساختار داده‌ها باید بررسی شود.

موارد الزامی:

۱. بررسی ابعاد دیتاست و نوع داده‌های هر ستون.

۲. شناسایی مقادیر گمشده در ستون‌های مختلف.

۳. شناسایی رکوردهای تکراری و بررسی تأثیر آن‌ها بر تحلیل.

۴. بررسی فاکتورهای لغوشده با استفاده از الگوی شماره فاکتور و سایر اطلاعات مرتبط.

۵. بررسی مقادیر منفی در Quantity و UnitPrice.

۶. بررسی تفاوت بین لغو سفارش، مرجوعی کالا و خطای ثبت داده.

۷. بررسی توزیع تعداد خریدها، قیمت محصولات و ارزش تراکنش‌ها.

۸. شناسایی مشتریان دارای رفتارهای غیرعادی و مقادیر پرت.

۹. بررسی بازه زمانی داده‌ها و میزان فعالیت مشتریان در دوره‌های مختلف.

۱۰. بررسی تعداد مشتریان یکتا و تعداد فاکتورهای معتبر.

خروجی این مرحله:

گزارش ممیزی داده شامل تعداد رکوردهای اولیه، رکوردهای تکراری، مقادیر گمشده، تراکنش‌های مشکوک، رکوردهای حذف‌شده، اصلاح‌شده و نگه‌داشته‌شده.

نکته: حذف تمام مقادیر منفی بدون بررسی معنای آن‌ها مجاز نیست. هر تصمیم پاک‌سازی باید مستند و قابل دفاع باشد.

---

۴. مرحله دوم: آماده‌سازی و پاک‌سازی داده‌ها (Data Preparation)

در این مرحله، داده‌ها برای تحلیل رفتار خرید مشتریان آماده می‌شوند.

موارد الزامی:

۱. مدیریت مقادیر گمشده، به‌خصوص CustomerID.

۲. حذف یا مدیریت رکوردهای تکراری با توجه به ماهیت تراکنش.

۳. تفکیک فاکتورهای خرید از فاکتورهای لغوشده.

۴. شناسایی و مدیریت مرجوعی‌ها و مقادیر منفی.

۵. شناسایی قیمت‌های نامعتبر یا غیرمنطقی.

۶. محاسبه ارزش تراکنش با استفاده از:

Revenue = Quantity × UnitPrice

۷. محاسبه فروش خالص با درنظرگرفتن خریدها و مرجوعی‌های معتبر.

۸. ایجاد یک مجموعه داده مناسب برای تحلیل رفتار مشتریان.

۹. تبدیل اطلاعات تراکنشی به اطلاعات سطح مشتری (Customer-Level Dataset).

۱۰. مستندسازی تمام تغییرات انجام‌شده.

نکته مهم: هدف این مرحله، تهیه داده‌ای است که رفتار واقعی خرید مشتریان را تا حد امکان منعکس کند؛ نه صرفاً حذف داده‌های غیرعادی.

---

۵. مرحله سوم: مهندسی ویژگی‌های رفتاری مشتریان (Feature Engineering)

در این مرحله، اطلاعات خام تراکنش‌ها به شاخص‌های رفتاری قابل تحلیل تبدیل می‌شوند.

حداقل ۱۵ ویژگی باید طراحی شود.

الف) ویژگی‌های RFM

۱. Recency: فاصله زمانی آخرین خرید معتبر مشتری تا تاریخ مرجع.

۲. Frequency: تعداد فاکتورهای خرید معتبر مشتری.

۳. Monetary: مجموع ارزش خالص خرید مشتری.

ب) ویژگی‌های ارزش اقتصادی مشتری

۴. Average Order Value (AOV): میانگین ارزش هر فاکتور خرید.

۵. Total Quantity: مجموع تعداد کالاهای خریداری‌شده.

۶. Average Quantity: میانگین تعداد کالا در هر سفارش.

۷. Product Diversity: تعداد محصولات متمایزی که مشتری خریداری کرده است.

۸. Purchase Days: تعداد روزهای متفاوتی که مشتری خرید داشته است.

ج) ویژگی‌های تغییر رفتار خرید

۹. Order Trend: میزان تغییر تعداد سفارش‌ها در دو بازه زمانی هم‌اندازه.

۱۰. Revenue Trend: میزان تغییر ارزش خالص خرید در دو بازه زمانی هم‌اندازه.

۱۱. Average Interpurchase Time: میانگین فاصله زمانی بین خریدهای متوالی.

۱۲. Return Rate: نسبت مرجوعی‌های معتبر به خریدهای مشتری، در صورت امکان محاسبه قابل اتکا.

۱۳. Cancellation Rate: نسبت سفارش‌های لغوشده به سفارش‌های ثبت‌شده، در صورت امکان تفکیک معتبر.

د) ویژگی‌های زمانی

۱۴. Customer Tenure: مدت زمان سپری‌شده از اولین خرید تا تاریخ مرجع.

۱۵. Purchase Frequency per Month: میانگین تعداد سفارش‌ها در ماه.

۱۶. Active Months: تعداد ماه‌هایی که مشتری در آن‌ها خرید داشته است.

۱۷. Purchase Regularity: میزان نظم یا پراکندگی زمانی خرید مشتری.

برای هر ویژگی باید موارد زیر مشخص شود:

- روش محاسبه
- مفهوم رفتاری
- اهمیت در تحلیل مشتری
- احتمال وجود مقادیر پرت
- محدودیت‌های محاسبه
- ارتباط آن با سایر ویژگی‌ها

تمام ویژگی‌ها باید فقط از اطلاعات در دسترس تا تاریخ مرجع استخراج شوند.

۶. مرحله چهارم: تحلیل توصیفی و اکتشافی رفتار مشتریان (Exploratory Data Analysis)

پیش از اجرای الگوریتم خوشه‌بندی، باید رفتار کلی مشتریان بررسی شود.

موارد الزامی:

۱. بررسی توزیع تعداد سفارش‌ها در بین مشتریان.

۲. بررسی توزیع ارزش خرید مشتریان.

۳. تحلیل Recency و Frequency و Monetary.

۴. بررسی رابطه بین تعداد سفارش‌ها و ارزش خرید.

۵. بررسی تغییرات خرید در ماه‌های مختلف.

۶. بررسی مشتریان با خریدهای پرتکرار و کم‌ارزش در مقابل مشتریان با خریدهای کم‌تعداد و پرارزش.

۷. بررسی میزان مرجوعی و لغو سفارش.

۸. مقایسه رفتار خرید مشتریان در کشورهای مختلف.

۹. شناسایی مشتریان دارای ارزش اقتصادی بالا و فعالیت خرید پایین.

۱۰. بررسی همبستگی میان ویژگی‌های رفتاری.

خروجی این مرحله:

نمودارها، جداول توصیفی و تحلیل اولیه از ساختار رفتاری مشتریان.

---

۷. مرحله پنجم: بخش‌بندی مشتریان با استفاده از خوشه‌بندی (Customer Segmentation)

از آنجا که دیتاست دارای برچسب آماده‌ای برای تعیین گروه رفتاری مشتریان نیست، از الگوریتم‌های یادگیری بدون نظارت (Unsupervised Learning) استفاده می‌شود.

هدف این مرحله، کشف گروه‌های طبیعی مشتریان بر اساس شباهت رفتار خرید آن‌هاست.

الگوریتم‌های پیشنهادی

۱. K-Means

۲. Hierarchical Clustering

۳. DBSCAN (در صورت تناسب ساختار داده)

مراحل اجرای خوشه‌بندی

۱. انتخاب ویژگی‌های مناسب رفتاری.

۲. بررسی همبستگی و حذف ویژگی‌های تکراری یا نامناسب.

۳. مدیریت مقادیر پرت.

۴. نرمال‌سازی یا استانداردسازی ویژگی‌ها.

۵. تعیین تعداد مناسب خوشه‌ها با روش‌هایی مانند Elbow و Silhouette Score.

۶. اجرای الگوریتم خوشه‌بندی.

۷. مقایسه کیفیت خوشه‌ها.

۸. بررسی ویژگی‌های آماری هر خوشه.

۹. تفسیر رفتاری و اقتصادی خوشه‌ها.

نمونه گروه‌های احتمالی

- مشتریان وفادار و ارزشمند
- مشتریان پرتکرار با ارزش خرید متوسط
- مشتریان کم‌فعالیت
- مشتریان با ارزش خرید بالا اما خریدهای پراکنده
- مشتریان با کاهش فعالیت خرید
- مشتریان تازه‌وارد یا دارای سابقه خرید محدود

این گروه‌ها صرفاً نمونه‌هایی از تفسیر احتمالی خوشه‌ها هستند و نباید پیش از مشاهده نتایج به‌عنوان خروجی قطعی الگوریتم در نظر گرفته شوند.

نکته مهم: خوشه‌بندی برای کشف الگوهای رفتاری استفاده می‌شود و به‌تنهایی مشخص نمی‌کند که یک مشتری قطعاً ریزش کرده یا در آینده ریزش خواهد کرد.

---

۸. مرحله ششم: محاسبه ارزش اقتصادی و فرصت بازاریابی مشتریان

پس از شناسایی گروه‌های رفتاری، باید برای هر مشتری دو امتیاز مستقل محاسبه شود.

این دو امتیاز از نظر مفهومی متفاوت هستند و نباید با یکدیگر اشتباه گرفته شوند.

الف) امتیاز ارزش اقتصادی مشتری (Customer Value Score)

هدف: مشخص‌کردن میزان اهمیت اقتصادی هر مشتری برای کسب‌وکار.

ویژگی‌های پیشنهادی:

- Monetary
- Frequency
- Average Order Value

مشتریانی که ارزش خرید و سابقه سفارش بالاتری دارند، معمولاً ارزش اقتصادی بیشتری برای شرکت ایجاد کرده‌اند.

ب) امتیاز فرصت بازاریابی (Marketing Opportunity Score)

هدف: شناسایی مشتریانی که با توجه به رفتار اخیرشان، فرصت بیشتری برای اقدامات بازاریابی و توجه مجدد دارند.

ویژگی‌های پیشنهادی:

- Recency
- Order Trend
- Revenue Trend
- Return Rate (اختیاری و در صورت اعتبار داده)

برای مثال:

- مشتری‌ای که در گذشته خریدهای ارزشمندی داشته اما اخیراً فاصله بین خریدهایش افزایش یافته، می‌تواند فرصت مناسبی برای بازاریابی باشد.
- مشتری‌ای که تعداد سفارش‌ها و مبلغ خرید او کاهش یافته، ممکن است نیازمند بررسی و توجه بیشتری باشد.
- مشتری‌ای که اخیراً خرید قابل‌توجهی داشته و روند خریدش مثبت است، لزوماً در اولویت کمپین بازگردانی قرار نمی‌گیرد.

جهت محاسبه این امتیازها باید:

۱. مقیاس ویژگی‌ها یکسان‌سازی شود.

۲. جهت اثر هر ویژگی مشخص شود.

۳. وزن ویژگی‌ها با منطق تحلیلی و مستند تعیین شود.

۴. روش محاسبه امتیازها شفاف باشد.

۵. تحلیل حساسیت نسبت به وزن‌ها انجام شود.

۶. نتایج دو امتیاز به‌صورت جداگانه قابل مشاهده باشند.

نکته: Marketing Opportunity به معنای احتمال قطعی ریزش نیست؛ بلکه شاخصی برای اولویت‌بندی فرصت‌های بازاریابی بر اساس رفتار مشاهده‌شده مشتری است.

---

۹. مرحله هفتم: اولویت‌بندی و انتخاب ۵۰۰ مشتری (Customer Prioritization)

فرض کنید شرکت فقط بودجه لازم برای اجرای کمپین روی ۵۰۰ مشتری را دارد.

هدف، انتخاب مشتریانی است که از نظر ارزش اقتصادی و فرصت بازاریابی، بیشترین اولویت را دارند.

الزامات:

۱. محاسبه Customer Value Score برای تمام مشتریان واجد شرایط.

۲. محاسبه Marketing Opportunity Score برای تمام مشتریان واجد شرایط.

۳. ترکیب دو امتیاز با یک روش شفاف و مستند.

۴. رتبه‌بندی مشتریان بر اساس امتیاز نهایی.

۵. انتخاب ۵۰۰ مشتری اول.

۶. بررسی توزیع ۵۰۰ مشتری منتخب در خوشه‌های مختلف.

۷. بررسی اینکه چه تعداد از مشتریان منتخب ارزش اقتصادی بالا و چه تعداد فرصت بازاریابی بالا دارند.

۸. ارائه فهرست مشتریان با شناسه ناشناس‌شده.

تحلیل ماتریسی مشتریان

با ترکیب دو امتیاز می‌توان مشتریان را در چهار گروه تحلیلی قرار داد:

- ارزش اقتصادی بالا + فرصت بازاریابی بالا
- ارزش اقتصادی بالا + فرصت بازاریابی پایین
- ارزش اقتصادی پایین + فرصت بازاریابی بالا
- ارزش اقتصادی پایین + فرصت بازاریابی پایین

این تقسیم‌بندی به مدیریت کمک می‌کند تا علاوه بر شناخت مشتریان ارزشمند، فرصت‌های بازاریابی را نیز بهتر شناسایی کند.

---

۱۰. مرحله هشتم: کشف قوانین انجمنی (Association Rule Mining)

در این مرحله، الگوهای هم‌خریدی محصولات بررسی می‌شوند تا مشخص شود چه محصولاتی معمولاً در یک فاکتور خریداری می‌شوند.

هدف، کشف روابط و الگوهای خرید برای طراحی پیشنهادهای محصول مرتبط و بهبود کمپین‌های بازاریابی است.

الگوریتم پیشنهادی

Apriori

در صورت نیاز می‌توان از FP-Growth نیز استفاده کرد.

مراحل اجرا

۱. تشکیل سبد خرید بر اساس InvoiceNo.

۲. شناسایی محصولات موجود در هر فاکتور معتبر.

۳. تبدیل اطلاعات به ساختار مناسب تحلیل قوانین انجمنی.

۴. استخراج مجموعه اقلام پرتکرار (Frequent Itemsets).

۵. استخراج قوانین انجمنی.

۶. بررسی معیارهای زیر:

- Support
- Confidence
- Lift

۷. شناسایی محصولات دارای ارتباط خرید.

۸. انتخاب قوانین معتبر و قابل تفسیر.

۹. طراحی پیشنهادهای محصول مرتبط بر اساس نتایج.

## شناخت کسبو کار و زمینه مسئله (Business Background)

### Business Background

فروشگاه‌های آنلاین با مجموعه‌ای از مشتریان مواجه هستند که از نظر الگوی خرید، میزان فعالیت، تعداد سفارش‌ها، ارزش خرید و تنوع محصولات مورد علاقه، رفتارهای متفاوتی دارند.

در یک فروشگاه آنلاین، همه مشتریان ارزش یکسانی برای کسب‌وکار ایجاد نمی‌کنند. برخی مشتریان به‌صورت مستمر خرید می‌کنند و سهم قابل‌توجهی در درآمد فروشگاه دارند. برخی دیگر سفارش‌های کمتری ثبت می‌کنند اما ارزش هر سفارش آن‌ها بالاست. گروهی از مشتریان نیز خریدهای پراکنده و نامنظم دارند یا در مقایسه با گذشته، فعالیت خرید کمتری از خود نشان می‌دهند.

از طرف دیگر، ممکن است مشتریانی وجود داشته باشند که سابقه خرید قابل‌توجهی داشته‌اند اما در دوره اخیر فعالیت آن‌ها کاهش یافته باشد. این مشتریان می‌توانند از نظر بازاریابی فرصت مناسبی برای برقراری ارتباط مجدد باشند.

همچنین شناخت محصولات مورد علاقه مشتریان و کالاهایی که معمولاً به‌صورت هم‌زمان خریداری می‌شوند، می‌تواند به فروشگاه کمک کند پیشنهادهای مرتبط‌تری ارائه دهد.بنابراین، مدیریت فروشگاه برای تصمیم‌گیری مؤثر به شناخت دقیق‌تری از ساختار مشتریان، رفتار خرید و ارزش اقتصادی آن‌ها نیاز دارد.

این مسئله ضرورت استفاده از روش‌های داده‌کاوی برای استخراج الگوهای پنهان موجود در سوابق تراکنش را ایجاد می‌کند.

## شرح مسئله کسبو کار (Business problem)

### Business Problem

مسئله اصلی کسب‌وکار، نبود شناخت کافی از ساختار رفتاری و اقتصادی مشتریان و در نتیجه دشواری در تصمیم‌گیری درباره نحوه تخصیص منابع بازاریابی است.

در وضعیت فعلی، اطلاعات تراکنش‌ها صرفاً سوابق خرید مشتریان را نشان می‌دهند؛ اما به‌تنهایی مشخص نمی‌کنند که:

- مشتریان فروشگاه چه گروه‌های رفتاری متفاوتی دارند؟
- کدام مشتریان بیشترین ارزش اقتصادی را برای فروشگاه ایجاد کرده‌اند؟
- کدام مشتریان با وجود ارزش اقتصادی قابل‌توجه، در دوره اخیر فعالیت خرید کمتری داشته‌اند؟
- کدام مشتریان فرصت مناسبی برای بازاریابی مجدد دارند؟
- کدام گروه‌های مشتریان به برنامه‌های حفظ مشتری نیاز دارند؟
- چه محصولاتی معمولاً در کنار یکدیگر خریداری می‌شوند؟
- با توجه به محدودیت بودجه، کدام مشتریان باید در اولویت کمپین بازاریابی قرار بگیرند؟

از آنجا که فروشگاه بودجه و منابع نامحدودی برای اجرای کمپین‌های بازاریابی ندارد، انتخاب مشتریان صرفاً بر اساس تعداد خرید یا مبلغ خرید گذشته نمی‌تواند تصمیم مناسبی باشد.برای مثال، مشتری‌ای که در گذشته خریدهای ارزشمندی داشته اما اخیراً فعالیت خرید او کاهش یافته است، ممکن است فرصت بازاریابی بیشتری نسبت به مشتری‌ای داشته باشد که اخیراً خریدهای مستمر و قابل‌توجهی انجام داده است.

بنابراین، مسئله کسب‌وکار به شناسایی هم‌زمان دو جنبه مهم نیاز دارد:

1. ارزش اقتصادی مشتری برای کسب‌وکار.
2. میزان فرصت موجود برای اقدامات بازاریابی درباره آن مشتری.

هدف پروژه، تبدیل سوابق تراکنش به اطلاعات تحلیلی و قابل استفاده برای تصمیم‌گیری مدیریتی است.

## مسئله اصلی و چالش داده ای (Business Challenge & Data Challenge)


### Business Challenge

چالش اصلی کسب‌وکار، شناسایی مشتریانی است که از نظر اقتصادی اهمیت دارند و می‌توانند فرصت مناسبی برای اقدامات بازاریابی ایجاد کنند.

با توجه به محدودیت بودجه، فروشگاه نمی‌تواند برای تمام مشتریان برنامه بازاریابی یکسانی اجرا کند. بنابراین لازم است مشتریان بر اساس رفتار خرید، ارزش اقتصادی و فرصت بازاریابی اولویت‌بندی شوند.

### Data Challenge

دیتاست Online Retail فاقد ستون یا برچسب آماده‌ای برای تعیین وضعیت مشتری از نظر ریزش، وفاداری، ارزش اقتصادی یا فرصت بازاریابی است.

بنابراین:
- نمی‌توان مستقیماً از یک ستون آماده برای تعیین گروه رفتاری مشتریان استفاده کرد.
- اطلاعات تراکنشی باید ابتدا به ویژگی‌های رفتاری در سطح مشتری تبدیل شوند.
- گروه‌های رفتاری مشتریان باید با استفاده از روش‌های خوشه‌بندی کشف شوند.
- ارزش اقتصادی مشتریان باید از روی شاخص‌های خرید محاسبه شود.
- فرصت بازاریابی باید با استفاده از شاخص‌های رفتاری و روند فعالیت مشتریان ارزیابی شود.

در نتیجه، رویکرد اصلی پروژه مبتنی بر یادگیری بدون نظارت، مهندسی ویژگی، تحلیل اقتصادی مشتریان و امتیازدهی تحلیلی است؛ نه آموزش یک مدل طبقه‌بندی برای پیش‌بینی ریزش.

نکته مهم این است که نبود برچسب ریزش به‌معنای نبود امکان تحلیل رفتار مشتری نیست. می‌توان بدون داشتن برچسب آماده، ساختار رفتاری مشتریان را کشف و وضعیت اقتصادی و فعالیت آن‌ها را ارزیابی کرد.

## شناسایی فرصت کسب و کار (Business Opportunity)

## Business Opportunity

تحلیل سوابق تراکنش فروشگاه می‌تواند فرصت‌های متعددی برای بهبود عملکرد کسب‌وکار ایجاد کند.

مهم‌ترین فرصت‌های قابل شناسایی عبارت‌اند از:

1. شناخت مشتریان ارزشمند:
   شناسایی مشتریانی که از نظر مبلغ خرید، تعداد سفارش و ارزش اقتصادی، سهم قابل‌توجهی در درآمد فروشگاه دارند.

2. شناسایی گروه‌های رفتاری:
   کشف گروه‌های مختلف مشتریان بر اساس شباهت در الگوی خرید و میزان فعالیت.

3. شناسایی فرصت‌های بازاریابی:
   شناسایی مشتریانی که سابقه خرید قابل‌توجهی دارند اما فعالیت اخیر آن‌ها کاهش یافته است.

4. بهبود تخصیص بودجه:
   انتخاب مشتریان مناسب برای کمپین بازاریابی بر اساس ترکیب ارزش اقتصادی و فرصت بازاریابی.

5. توسعه فروش محصولات:
   شناسایی محصولات دارای ارتباط خرید با استفاده از قوانین انجمنی و طراحی پیشنهادهای مکمل.

6. طراحی راهبردهای اختصاصی:
   ارائه پیشنهادهای بازاریابی متناسب با ویژگی‌های هر گروه مشتری به‌جای اجرای یک کمپین یکسان برای همه.

7. افزایش بهره‌وری بازاریابی:
   کاهش هزینه‌های ناشی از ارتباطات عمومی و تخصیص هدفمند منابع به مشتریان دارای اولویت بالاتر.

بنابراین، داده‌های تراکنشی فروشگاه می‌توانند از یک منبع صرفاً ثبت‌کننده فروش به ابزاری برای شناخت مشتری و پشتیبانی از تصمیم‌های تجاری تبدیل شوند.



## هدف اصلی کسب و کار (Primary Business Objective)


### Primary Business Objective

هدف اصلی کسب‌وکار، ایجاد یک چارچوب تحلیلی برای شناخت رفتار مشتریان، بخش‌بندی آن‌ها بر اساس شباهت رفتاری، ارزیابی ارزش اقتصادی و شناسایی فرصت‌های بازاریابی است.

این چارچوب باید به مدیریت کمک کند:

- ساختار واقعی مشتریان فروشگاه را شناسایی کند.
- گروه‌های مختلف مشتریان را از نظر رفتار خرید و ارزش اقتصادی مقایسه کند.
- مشتریان ارزشمند را از مشتریان کم‌ارزش‌تر تفکیک کند.
- فرصت‌های بازاریابی و ارتباط مجدد با مشتریان را شناسایی کند.
- مشتریان را بر اساس اولویت تجاری رتبه‌بندی کند.
- با توجه به محدودیت بودجه، ۵۰۰ مشتری مناسب را برای کمپین انتخاب کند.
- برای هر گروه مشتری، راهبرد بازاریابی متناسب طراحی کند.
- از الگوهای هم‌خریدی محصولات برای ارائه پیشنهادهای مرتبط استفاده کند.

هدف نهایی، بهبود کیفیت تصمیم‌گیری بازاریابی و افزایش بهره‌وری منابع شرکت از طریق استفاده از شواهد استخراج‌شده از داده‌های خرید است.

## اهداف فرعی کسبو کار 


### Business Objectives

اهداف فرعی کسب‌وکار به شرح زیر تعریف می‌شوند:

#### 1 – شناخت ساختار مشتریان
شناسایی الگوهای مختلف رفتار خرید و کشف گروه‌های طبیعی مشتریان بر اساس ویژگی‌های تراکنشی و رفتاری.

#### 2 – بخش‌بندی مشتریان
تقسیم مشتریان به گروه‌های رفتاری قابل تفسیر با استفاده از الگوریتم‌های خوشه‌بندی.

#### 3 – ارزیابی ارزش اقتصادی
اندازه‌گیری ارزش اقتصادی مشتریان بر اساس شاخص‌هایی مانند مبلغ خرید، تعداد سفارش‌ها و میانگین ارزش سفارش.

#### 4 – شناسایی فرصت‌های بازاریابی
شناسایی مشتریانی که با توجه به سابقه خرید و تغییرات فعالیت اخیر، فرصت مناسبی برای اقدامات بازاریابی دارند.

#### 5 – اولویت‌بندی مشتریان
ترکیب اطلاعات ارزش اقتصادی و فرصت بازاریابی برای رتبه‌بندی مشتریان و انتخاب ۵۰۰ مشتری اولویت‌دار.

#### 6 – طراحی راهبردهای بازاریابی
تعیین راهبردهای مناسب برای گروه‌های مختلف مشتریان با توجه به رفتار خرید و ارزش اقتصادی آن‌ها.

#### 7 – کشف الگوهای هم‌خریدی محصولات
شناسایی روابط بین محصولات و استخراج قوانین انجمنی برای طراحی پیشنهادهای محصول مرتبط.

#### 8 – بهبود تصمیم‌گیری مدیریتی
ارائه گزارش‌ها، جداول و شاخص‌های قابل تفسیر برای پشتیبانی از تصمیم‌های مدیریت فروشگاه در زمینه بازاریابی و تخصیص منابع.

## سئوالات کسبو کار (Business Questions)

## Business Questions

برای پاسخ‌گویی به مسئله کسب‌وکار، سؤالات زیر مطرح می‌شوند:

### A. شناخت ساختار مشتریان

1. مشتریان فروشگاه از نظر رفتار خرید چه تفاوت‌هایی با یکدیگر دارند؟
2. چه الگوهای متفاوتی از نظر تعداد سفارش، ارزش خرید و زمان‌بندی خرید در میان مشتریان وجود دارد؟
3. آیا می‌توان مشتریان را بر اساس شباهت رفتاری به گروه‌های مشخص تقسیم کرد؟
4. چه گروه‌هایی از نظر تعداد مشتری، میزان خرید و سهم درآمد اهمیت بیشتری دارند؟
5. آیا گروهی از مشتریان وجود دارد که تعداد سفارش بالایی اما ارزش متوسط هر سفارش پایینی داشته باشد؟
6. آیا گروهی از مشتریان وجود دارد که تعداد سفارش آن‌ها کم اما ارزش اقتصادی خریدشان بالا باشد؟

### B. شناخت ارزش اقتصادی مشتریان

7. کدام مشتریان بیشترین ارزش اقتصادی را برای فروشگاه ایجاد کرده‌اند؟
8. چه رابطه‌ای بین تعداد سفارش‌ها و ارزش اقتصادی مشتریان وجود دارد؟
9. آیا مشتریان پرتکرار الزاماً ارزش اقتصادی بیشتری نسبت به سایر مشتریان دارند؟
10. کدام مشتریان با وجود تعداد سفارش محدود، ارزش خرید بالایی ایجاد کرده‌اند؟
11. چه گروه‌هایی سهم بیشتری از درآمد خالص فروشگاه را تشکیل می‌دهند؟
12. آیا ارزش اقتصادی مشتریان در خوشه‌های مختلف تفاوت معناداری دارد؟

### C. شناخت تغییرات رفتار خرید

13. کدام مشتریان در مقایسه با دوره قبلی کاهش فعالیت خرید داشته‌اند؟
14. آیا کاهش تعداد سفارش‌ها با کاهش ارزش خرید همراه بوده است؟
15. کدام مشتریان سابقه خرید ارزشمند دارند اما در دوره اخیر فعالیت کمتری داشته‌اند؟
16. آیا مشتریان با خریدهای نامنظم از نظر ارزش اقتصادی با مشتریان منظم تفاوت دارند؟
17. چه رابطه‌ای بین فاصله زمانی آخرین خرید و ارزش اقتصادی مشتری وجود دارد؟

### D. شناسایی فرصت‌های بازاریابی

18. کدام مشتریان با توجه به ارزش اقتصادی گذشته و تغییرات فعالیت اخیر، فرصت بیشتری برای بازاریابی دارند؟
19. آیا مشتریان ارزشمند با فعالیت خرید پایین می‌توانند در اولویت کمپین بازاریابی قرار بگیرند؟
20. کدام گروه‌های مشتریان بیشتر به برنامه‌های حفظ مشتری نیاز دارند؟
21. کدام مشتریان برای بازاریابی مجدد و کدام مشتریان برای توسعه خرید مناسب‌تر هستند؟
22. چگونه می‌توان ارزش اقتصادی را از فرصت بازاریابی متمایز کرد؟

### E. تخصیص منابع بازاریابی

23. با توجه به محدودیت بودجه، چگونه می‌توان ۵۰۰ مشتری را برای اجرای کمپین انتخاب کرد؟
24. چه ترکیبی از ارزش اقتصادی و فرصت بازاریابی باید در اولویت انتخاب مشتریان قرار بگیرد؟
25. مشتریان منتخب از نظر خوشه رفتاری چه توزیعی دارند؟
26. آیا انتخاب مشتریان صرفاً بر اساس ارزش خرید گذشته می‌تواند موجب نادیده‌گرفتن فرصت‌های بازاریابی شود؟

### F. تحلیل الگوهای خرید محصولات

27. چه محصولاتی معمولاً در یک فاکتور خریداری می‌شوند؟
28. کدام محصولات می‌توانند به‌عنوان پیشنهاد مکمل برای یکدیگر استفاده شوند؟
29. آیا الگوهای هم‌خریدی محصولات در گروه‌های مختلف مشتریان متفاوت است؟
30. چگونه می‌توان از قوانین انجمنی برای طراحی پیشنهادهای محصول مرتبط استفاده کرد؟

این سؤالات چارچوب تصمیم‌گیری کسب‌وکار را مشخص می‌کنند و مبنای تعیین اهداف داده‌کاوی خواهند بود.

## تصمیم های مورد انتظار کسبو کار (Business Decisions)

## Business Decisions

نتایج پروژه باید امکان اتخاذ تصمیم‌های زیر را برای مدیریت فراهم کنند:

1. تعیین اینکه مشتریان فروشگاه به چه گروه‌های رفتاری تقسیم می‌شوند.

2. تشخیص اینکه کدام گروه‌ها بیشترین ارزش اقتصادی را برای فروشگاه ایجاد می‌کنند.

3. تعیین اینکه کدام مشتریان ارزش اقتصادی بالایی دارند اما فعالیت اخیر آن‌ها کاهش یافته است.

4. تصمیم‌گیری درباره اینکه کدام گروه‌ها باید در برنامه‌های حفظ مشتری قرار بگیرند.

5. تعیین اینکه کدام مشتریان برای توسعه خرید و فروش محصولات بیشتر مناسب هستند.

6. انتخاب ۵۰۰ مشتری اولویت‌دار برای اجرای کمپین بازاریابی.

7. تعیین نوع ارتباط و پیشنهاد مناسب برای هر گروه مشتری.

8. انتخاب محصولات مکمل بر اساس الگوهای هم‌خریدی.

9. تخصیص بودجه بازاریابی بر اساس ارزش اقتصادی و فرصت بازاریابی.

10. ارزیابی اقتصادی کمپین و بررسی امکان اجرای آن.

11. تعیین اینکه کدام گروه‌های مشتریان به سرمایه‌گذاری بیشتر در برنامه‌های وفاداری نیاز دارند.

12. شناسایی فرصت‌های بهبود تجربه خرید و توسعه فروش در گروه‌های مختلف مشتریان.

در نتیجه، خروجی تحلیل باید به تصمیم عملی منجر شود و صرفاً به تولید نمودار یا تشکیل خوشه‌های عددی محدود نماند.

## اهداف داده کاوی (Data Mining Goals)

### Data Mining Goals

برای پاسخ‌گویی به اهداف کسب‌وکار، اهداف داده‌کاوی زیر تعریف می‌شوند:

#### 1 – آماده‌سازی داده‌های رفتاری
تبدیل اطلاعات تراکنشی خام به داده‌های معتبر در سطح مشتری و استخراج ویژگی‌های رفتاری قابل استفاده در تحلیل.

#### 2 – مهندسی ویژگی
محاسبه شاخص‌های رفتاری و زمانی مانند Recency، Frequency، Monetary، Average Order Value، Product Diversity، Purchase Regularity و شاخص‌های تغییر رفتار خرید.

#### 3 – تحلیل اکتشافی رفتار مشتریان
بررسی توزیع ویژگی‌ها، روابط میان متغیرها، الگوهای خرید و تفاوت‌های رفتاری مشتریان.

#### 4 – کشف گروه‌های رفتاری
استفاده از الگوریتم‌های خوشه‌بندی مانند K-Means و Hierarchical Clustering برای شناسایی گروه‌های طبیعی مشتریان.

#### 5 – ارزیابی و تفسیر خوشه‌ها
مقایسه کیفیت خوشه‌بندی با استفاده از معیارهای مناسب و تحلیل ویژگی‌های رفتاری و اقتصادی هر خوشه.

#### 6 – محاسبه ارزش اقتصادی مشتری
ایجاد یک شاخص ترکیبی بر اساس ویژگی‌هایی مانند Monetary، Frequency و Average Order Value برای ارزیابی ارزش اقتصادی مشتریان.

#### 7 – محاسبه فرصت بازاریابی
ایجاد یک شاخص تحلیلی مستقل بر اساس Recency، Order Trend، Revenue Trend و در صورت اعتبار، Return Rate برای شناسایی فرصت‌های بازاریابی.

#### 8 – اولویت‌بندی مشتریان
ترکیب امتیاز ارزش اقتصادی و فرصت بازاریابی با یک روش شفاف و مستند و رتبه‌بندی مشتریان.

#### 9 – انتخاب مشتریان هدف
انتخاب ۵۰۰ مشتری اولویت‌دار و بررسی توزیع آن‌ها در خوشه‌های مختلف.

#### 10 – استخراج قوانین انجمنی
استفاده از الگوریتم Apriori برای شناسایی مجموعه اقلام پرتکرار و استخراج قوانین هم‌خریدی محصولات.

#### 11 – تحلیل تفاوت الگوهای خرید
بررسی تفاوت الگوهای هم‌خریدی در گروه‌های مختلف مشتریان، در صورت کافی‌بودن داده‌ها.

#### 12 – ارزیابی پایداری نتایج
بررسی پایداری خوشه‌ها، امتیازها و رتبه‌بندی مشتریان در تاریخ‌های مرجع مختلف با رعایت محدودیت اطلاعات زمانی.

#### 13 – تولید خروجی قابل استفاده
تولید جداول، نمودارها، گزارش‌ها و فهرست مشتریان منتخب برای پشتیبانی از تصمیم‌های بازاریابی.

این ارتباط تضمین می‌کند که هر فعالیت داده‌کاوی در راستای پاسخ‌گویی به یک نیاز مشخص کسب‌وکار انجام شود.

## تعریف مفاهیم کلیدی  پروژه (Key Business Definitions)


### Key Business Definitions

برای جلوگیری از ابهام در تحلیل، مفاهیم کلیدی پروژه به شکل زیر تعریف می‌شوند:

#### 1. Customer Segmentation
فرایند تقسیم مشتریان به گروه‌هایی بر اساس شباهت ویژگی‌های رفتاری و اقتصادی آن‌ها.

#### 2. Customer Economic Value
میزان ارزش اقتصادی ایجادشده توسط مشتری در بازه تاریخی مورد بررسی که با استفاده از شاخص‌های خرید محاسبه می‌شود.

این مفهوم در پروژه به معنای سود خالص یا ارزش طول عمر واقعی مشتری نیست؛ زیرا اطلاعات کامل هزینه‌های عملیاتی، حاشیه سود و هزینه جذب مشتری در دیتاست موجود نیست.

#### 3. Marketing Opportunity
میزان اولویت تحلیلی یک مشتری برای اقدامات بازاریابی با توجه به ارزش خرید گذشته، فاصله زمانی آخرین خرید و تغییرات رفتار خرید.

این شاخص احتمال قطعی خرید مجدد یا ریزش مشتری را نشان نمی‌دهد.

#### 4. Customer Activity
میزان فعالیت خرید مشتری در یک دوره مشخص که با استفاده از تعداد سفارش‌ها، روزهای خرید و شاخص‌های زمانی اندازه‌گیری می‌شود.

#### 5. Active Customer
مشتری‌ای که در بازه زمانی تعریف‌شده، حداقل یک خرید معتبر داشته باشد.

#### 6. Low-Activity Customer
مشتری‌ای که در مقایسه با معیار یا دوره مرجع تعیین‌شده، سطح فعالیت خرید پایین‌تری دارد.

#### 7. High-Value Customer
مشتری‌ای که بر اساس شاخص ارزش اقتصادی محاسبه‌شده، در گروه بالاتری از ارزش خرید قرار می‌گیرد.

#### 8. High Marketing Opportunity Customer
مشتری‌ای که بر اساس شاخص فرصت بازاریابی، اولویت بالاتری برای بررسی و اجرای اقدامات بازاریابی دارد.

#### 9. Purchase Pattern
الگوی رفتاری مشتری در زمینه تعداد سفارش‌ها، مبلغ خرید، زمان خرید و نوع محصولات خریداری‌شده.

#### 10. Cross-Selling
ارائه محصولات مکمل یا مرتبط به مشتری بر اساس الگوهای خرید و روابط کشف‌شده میان محصولات.

#### 11. Customer Prioritization
فرایند رتبه‌بندی مشتریان بر اساس ترکیب ارزش اقتصادی و فرصت بازاریابی با هدف تخصیص منابع محدود. مفاهیم بالا تعاریف تحلیلی پروژه هستند. تعیین حدود عددی هر گروه و روش محاسبه دقیق آن‌ها در مراحل بعدی و پس از بررسی داده‌ها انجام خواهد شد.

## چهار چوب تحلیلی و رویکرد حل مسئله (Analytical Approach)


### Analytical Approach

با توجه به ماهیت مسئله و نبود برچسب آماده برای گروه‌های رفتاری مشتریان، رویکرد تحلیلی پروژه بر پایه یادگیری بدون نظارت و تحلیل شاخص‌های اقتصادی و رفتاری طراحی می‌شود.

سه محور اصلی تحلیل عبارت‌اند از:

#### محور اول: Customer Segmentation

هدف:
کشف گروه‌های طبیعی مشتریان بر اساس شباهت رفتار خرید.

روش:
- استخراج ویژگی‌های رفتاری در سطح مشتری.
- انتخاب ویژگی‌های مناسب.
- استانداردسازی داده‌ها.
- اجرای الگوریتم‌های خوشه‌بندی.
- ارزیابی کیفیت خوشه‌ها.
- تفسیر رفتاری و اقتصادی هر خوشه.

الگوریتم‌های مورد بررسی:
- K-Means
- Hierarchical Clustering
- DBSCAN در صورت تناسب ساختار داده‌ها

#### محور دوم: Customer Economic Value and Marketing Opportunity

هدف:
ارزیابی مستقل ارزش اقتصادی و فرصت بازاریابی هر مشتری.

روش:
- محاسبه Customer Value Score بر اساس شاخص‌های اقتصادی و خرید.
- محاسبه Marketing Opportunity Score بر اساس فعالیت اخیر و تغییرات رفتار خرید.
- تعیین جهت اثر ویژگی‌ها.
- استانداردسازی شاخص‌ها.
- تعیین وزن‌های تحلیلی.
- بررسی حساسیت امتیازها نسبت به وزن‌ها.

در این محور از مدل طبقه‌بندی استفاده نمی‌شود و امتیازها به‌عنوان شاخص‌های تحلیلی و تصمیم‌یار تعریف می‌شوند.

#### محور سوم: Association Rule Mining

هدف:
کشف روابط میان محصولات و الگوهای هم‌خریدی.

روش:
- تشکیل سبد خرید بر اساس فاکتورهای معتبر.
- استخراج مجموعه اقلام پرتکرار.
- اجرای الگوریتم Apriori.
- محاسبه Support، Confidence و Lift.
- انتخاب قوانین قابل تفسیر.
- طراحی پیشنهادهای محصول مرتبط.

#### نحوه ترکیب سه محور

خروجی خوشه‌بندی برای شناخت ساختار رفتاری مشتریان استفاده می‌شود.

خروجی امتیازدهی برای تعیین ارزش اقتصادی و فرصت بازاریابی و اولویت‌بندی مشتریان به کار می‌رود.

خروجی قوانین انجمنی برای پیشنهاد محصولات مرتبط و طراحی کمپین‌های هدفمند استفاده می‌شود.

این سه محور مکمل یکدیگر هستند؛ اما از نظر هدف تحلیلی مستقل باقی می‌مانند.

## معیار های موفقیت کسبو کار(Business Success Criteria)


### Business Success Criteria

موفقیت پروژه از نظر کسب‌وکار بر اساس معیارهای زیر ارزیابی می‌شود:

#### 1. شناخت ساختار مشتریان
امکان شناسایی و تفسیر گروه‌های رفتاری متفاوت بر اساس داده‌های خرید.

#### 2. تفکیک ارزش اقتصادی
امکان شناسایی مشتریان با ارزش اقتصادی بالا و مقایسه سهم اقتصادی گروه‌های مختلف.

#### 3. شناسایی فرصت بازاریابی
امکان شناسایی مشتریانی که با توجه به سابقه خرید و تغییرات فعالیت، اولویت مناسبی برای بازاریابی دارند.

#### 4. اولویت‌بندی مشتریان
تولید یک فهرست رتبه‌بندی‌شده از ۵۰۰ مشتری واجد شرایط بر اساس معیارهای مشخص و قابل دفاع.

#### 5. قابلیت استفاده مدیریتی
امکان تبدیل نتایج خوشه‌بندی و امتیازدهی به پیشنهادهای اجرایی برای واحد بازاریابی.

#### 6. کشف روابط محصولات
استخراج قوانین انجمنی معتبر و قابل استفاده برای پیشنهاد محصولات مرتبط.

#### 7. پایداری نتایج
برخورداری از نتایج نسبتاً پایدار در تاریخ‌های مرجع مختلف و حساسیت قابل قبول نسبت به تغییر پارامترها و وزن‌ها.

#### 8. قابلیت دفاع از تصمیم‌ها
امکان توضیح اینکه چرا یک مشتری در یک خوشه قرار گرفته یا چرا در اولویت بالاتری برای کمپین انتخاب شده است.

#### 9. ارزیابی اقتصادی کمپین
امکان برآورد هزینه، درآمد فرضی، سود احتمالی و نقطه سربه‌سر با استفاده از فرض‌های شفاف.

#### 10. قابلیت بازتولید
امکان اجرای مجدد فرایند تحلیل و تولید خروجی‌های مشابه با داده‌ها و تنظیمات یکسان.

## محدودیت ها و فرضیات کسبو کار (Business Constraints & Assumptions)

### Business Constraints

محدودیت‌های اصلی پروژه عبارت‌اند از:

1. نبود برچسب آماده برای تعیین وضعیت ریزش یا وفاداری مشتریان.
2. محدودبودن اطلاعات به سوابق تراکنشی ثبت‌شده.
3. محدودبودن بازه زمانی داده‌ها به دسامبر ۲۰۱۰ تا دسامبر ۲۰۱۱.
4. نبود اطلاعات کامل درباره هزینه‌های عملیاتی و حاشیه سود محصولات.
5. نبود اطلاعات مربوط به هزینه جذب هر مشتری.
6. نبود اطلاعات مستقیم درباره رضایت، ترجیحات اعلام‌شده و تجربه مشتری.
7. وجود مقادیر گمشده، رکوردهای تکراری، فاکتورهای لغوشده و مقادیر غیرعادی.
8. احتمال وجود سوگیری ناشی از محدودبودن بازه زمانی مشاهده.
9. محدودیت بودجه بازاریابی و ضرورت انتخاب حداکثر ۵۰۰ مشتری.
10. احتمال ناپایداری برخی خوشه‌ها و شاخص‌ها در اثر انتخاب ویژگی‌ها و پارامترهای مختلف.


### Business Assumptions

فرضیات اولیه پروژه عبارت‌اند از:

1. هر CustomerID نماینده یک شناسه مشتری در داده‌های تراکنشی است.
2. فاکتورهای خرید معتبر می‌توانند برای استخراج رفتار خرید مشتریان استفاده شوند.
3. تاریخ ثبت تراکنش‌ها نمایانگر زمان ثبت خرید در دیتاست است.
4. ارزش خرید خالص مشتری را می‌توان با درنظرگرفتن خریدها و مرجوعی‌های معتبر محاسبه کرد.
5. شاخص‌های RFM می‌توانند اطلاعات مفیدی درباره رفتار خرید مشتریان ارائه دهند.
6. کاهش فعالیت خرید در مقایسه با یک دوره مرجع می‌تواند نشانه‌ای از تغییر رفتار باشد؛ اما به‌تنهایی اثبات‌کننده ریزش نیست.
7. ارزش اقتصادی گذشته می‌تواند یکی از معیارهای تصمیم‌گیری درباره اولویت بازاریابی باشد.
8. ترکیب ارزش اقتصادی و فرصت بازاریابی می‌تواند مبنای مناسب‌تری برای اولویت‌بندی نسبت به استفاده از یک شاخص منفرد باشد.
9. الگوهای هم‌خریدی محصولات می‌توانند برای طراحی پیشنهادهای مرتبط مورد استفاده قرار گیرند.
10. نتایج تحلیل باید با توجه به محدودیت زمانی و کیفیت داده‌ها تفسیر شوند.

این فرضیات در مراحل Data Understanding و Data Preparation بررسی و در صورت نیاز اصلاح خواهند شد.

### Expected Business Value

اجرای این پروژه می‌تواند ارزش‌های زیر را برای فروشگاه ایجاد کند:

#### ارزش اقتصادی
- شناسایی مشتریانی که سهم بیشتری در درآمد فروشگاه دارند.
- کمک به تخصیص منابع بازاریابی به مشتریان دارای اولویت بالاتر.

- طراحی کمپین‌های متناسب با گروه‌های مختلف مشتریان.
- شناسایی مشتریان دارای فرصت بازاریابی مجدد.
- بهبود هدف‌گیری مشتریان.
- کاهش وابستگی به تبلیغات یکسان برای تمام مشتریان.

- تبدیل داده‌های خام تراکنشی به اطلاعات رفتاری.

- ایجاد مبنایی داده‌محور برای برنامه‌ریزی بازاریابی.

## تعریف خروجی نهایی پروژه(Final Business Deliverable)

### Final Business Deliverable

خروجی نهایی پروژه، یک سامانه تحلیلی پشتیبان تصمیم‌گیری بازاریابی است که اطلاعات تراکنشی را به اطلاعات قابل استفاده در تصمیم‌های مدیریتی تبدیل می‌کند.

این سامانه باید قابلیت‌های زیر را فراهم کند:

1. نمایش ساختار رفتاری مشتریان و گروه‌های شناسایی‌شده.
2. نمایش شاخص‌های رفتاری و اقتصادی هر مشتری.
3. ارائه Customer Value Score برای ارزیابی ارزش اقتصادی.
4. ارائه Marketing Opportunity Score برای ارزیابی فرصت بازاریابی.
5. نمایش ارتباط میان ارزش اقتصادی و فرصت بازاریابی در قالب ماتریس چهارگانه.
6. ارائه فهرست رتبه‌بندی‌شده ۵۰۰ مشتری منتخب.
7. نمایش توزیع مشتریان منتخب در خوشه‌های رفتاری.
8. ارائه پیشنهادهای بازاریابی متناسب با هر گروه.
9. نمایش قوانین انجمنی و پیشنهادهای محصولات مرتبط.
10. ارائه برآورد اقتصادی کمپین و نقطه سربه‌سر.
11. ارائه گزارش کیفیت و پایداری نتایج.

در نهایت، مدیریت باید بتواند با استفاده از خروجی این سامانه به سه سؤال اصلی پاسخ دهد:

- چه مشتریانی از نظر اقتصادی برای فروشگاه اهمیت بیشتری دارند؟
- کدام مشتریان در حال حاضر فرصت مناسب‌تری برای اقدامات بازاریابی دارند؟
- برای هر گروه از مشتریان چه اقدام بازاریابی و چه پیشنهاد محصولی مناسب‌تر است؟

بنابراین، موفقیت نهایی پروژه در توانایی تبدیل داده‌های تراکنشی به تصمیم‌های بازاریابی قابل دفاع و مبتنی بر شواهد تعریف می‌شود.

## جمع بندی مرحله شناخت کسبو کار(Business Undrestanding Summary)

### Business Understanding Summary

در این مرحله، مسئله اصلی کسب‌وکار، نیازهای مدیریتی، اهداف تجاری، اهداف داده‌کاوی، ذی‌نفعان، محدودیت‌ها و خروجی مورد انتظار پروژه مشخص شدند.

بر اساس شناخت انجام‌شده، مسئله اصلی فروشگاه صرفاً شناسایی مشتریان غیرفعال یا پیش‌بینی ریزش نیست؛ بلکه شناخت ساختار رفتاری مشتریان، ارزیابی ارزش اقتصادی آن‌ها و شناسایی فرصت‌های مناسب بازاریابی است.

برای حل این مسئله، سه محور تحلیلی مستقل اما مکمل تعریف شد:

1. Customer Segmentation:
   برای کشف گروه‌های طبیعی مشتریان بر اساس شباهت رفتار خرید.

2. Customer Economic Value and Marketing Opportunity:
   برای ارزیابی مستقل ارزش اقتصادی و فرصت بازاریابی و اولویت‌بندی مشتریان.

3. Association Rule Mining:
   برای کشف الگوهای هم‌خریدی محصولات و طراحی پیشنهادهای مرتبط.

در نهایت، خروجی این فرایند باید امکان شناسایی گروه‌های رفتاری، تشخیص مشتریان ارزشمند، انتخاب ۵۰۰ مشتری اولویت‌دار و طراحی راهبردهای بازاریابی متناسب با هر گروه را فراهم کند.

با مشخص‌شدن اهداف و محدوده پروژه، در مرحله بعد ساختار و کیفیت دیتاست Online Retail بررسی خواهد شد تا مشخص شود اطلاعات موجود تا چه اندازه امکان تحقق اهداف تعریف‌شده را فراهم می‌کنند.
