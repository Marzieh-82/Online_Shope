```python

from pathlib import Path
import pandas as pd

def find_csv(filename):
    filename = filename.lower()
    current = Path.cwd().resolve()


    for root in [current, *current.parents]:
        matches = [
            p for p in root.rglob("*")
            if p.is_file() and p.name.lower() == filename
        ]

        if matches:
            
            matches.sort(key=lambda p: len(p.parts))
            if len(matches) > 1:
                print("هشدار: چند فایل با این نام پیدا شد:")
                for p in matches:
                    print(" -", p)

            selected = matches[0]
            print("فایل پیدا شد:", selected)
            return selected

    raise FileNotFoundError(
        f"فایل {filename} پیدا نشد. مسیر جست‌وجو: {current}"
    )



customer_path = find_csv("top500_customers.csv")
customer = pd.read_csv(customer_path)

print("Shape:", customer.shape)
display(customer.head(500))
```
